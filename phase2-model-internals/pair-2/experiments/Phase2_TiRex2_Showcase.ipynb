# How TiRex-2 responds to changing weather

**Phase II · Austria INCA 15 km · 405 cases**

The 32-hour patch leaves a measurable imprint on forecast sensitivity. Once colder temperatures enter the history, the forecast adjusts quickly, while large cooling and warming remain damped. Before a front, the prediction interval sits too high to capture much of the observed cooling.

| Question | Main result |
| --- | --- |
| Does patch position matter when every hour stays observed? | The newest two slots are **1.75–2.30×** as sensitive as the average slot under value changes, even with the newest patch removed. |
| How much of a large change does the median forecast capture? | At +24 h, mean q50 changes by **−0.93 °C** in strong cooling and **+1.20 °C** in strong warming. Observed changes are **−6.67 °C** and **+5.98 °C**. |
| Where do the intervals miss before a front? | At +12 h, **70%** of outcomes fall below q10, and mean median bias is **+5.59 °C**. |

Here **q50** is the forecast median and **q10–q90** is the central 80% prediction interval. Lead +1 h is the first forecast hour, and lag 1 is the last observed hour. Time relative to local front passage is shown separately from forecast lead. The figures and results are embedded for viewing on open.

## 1. The newest slots matter across the 32-hour grid

Raw gradients are strongest in the newest history. Folding by position shows a different comparison. Within older patches, the newest slots repeatedly account for a larger share of sensitivity. The bottom panels below show why raw magnitude and within-patch share tell different parts of the story.

## 2. The position effect is already visible at the embedding

In the 24 pulse and embedding cases, equal changes at the scaled embedding input moved the token about **2.25×** as much for the newest two slots as for the average slot. Presence indicators were held fixed. In the same cases, a broad cooling pulse produced a smooth, lead-dependent response as it crossed two patch boundaries. Several individually sensitive hours can combine into a smooth change in the forecast.



*Dashed vertical lines mark the patch boundaries. The center panel isolates the embedding by changing one scaled slot at a time.*

## 3. The forecast adjusts to colder history and damps the ongoing trend

The curves align 24 local passages at each issuance offset. Forecasts issued before passage often project warmth into the front. By six and twelve hours after passage, q50 starts much closer to the observed cold level and predicts a shallower trough and rebound.



*The vertical line marks local passage. Last-value persistence and the extrapolated recent slope provide reference trajectories.*



*Each cohort has 48 cases. The groups are defined by observed +24 h change, so they describe the forecast response to realized extremes.*

### Older temperatures still shape the trajectory

At issuance six hours after passage, raising pre-front temperatures by 3 °C while fixing the recent cold segment increased the mean forecast level by **0.85 °C** and the forecast swing by **0.91 °C**. Removing the recent slope while fixing the last observation shifted the mean by **+0.06 °C** and the swing by **−0.19 °C**. The mean-level interval for slope removal included zero. Forecast swing is q50 at +24 h minus q50 at +1 h.



*Each scatter point is one post-passage case. The paired design holds the other inputs fixed within each comparison.*

### What the recurrent maps show

In the illustrated case, realized sLSTM memory is strongest near the source–receiver diagonal, and signed mixing in the mLSTM varies by head. These are 32-hour token maps for one case. Assigning the changes in forecast level or swing to a specific gate or head would require targeted internal interventions across cases.

## 4. Before passage, forecast intervals sit too warm

At +12 h, pre-front starts had a **+5.59 °C** q50 bias. **70%** of observations fell below q10, and only **30%** were inside the interval. The pre-front interval averaged **7.12 °C** wide versus **6.17 °C** for stable starts. The warm center and predominantly cold-side misses point to interval placement as the main issue at this lead. Coverage rose to **61%** for starts during passage and **79%** after passage.



*Passage phases are defined by issuance time. Each curve therefore tracks the same group across forecast leads.*



*The two groups have broadly similar lag profiles. Recent observations have large derivatives in both groups.*

With the final 24 observed hours fixed, four abrupt historical drops widened the mean band by **1.47 °C** at 9 °C amplitude and **0.91 °C** at 15 °C. Fading the endings reduced the widening, and the sweep means still declined at the largest amplitude. The logged differencing branch stayed fixed across the sweep.



*Solid lines use faded endings and dashed lines use abrupt endings. Amplitude zero is the untouched history.*

## 5. The model responds to covariates, with little timing benefit at +12 h

Across 40 pre-front cases in 12 blocks, all six +12 h absolute-error contrasts between the correctly timed inputs and their 24-hour-shifted controls were within **0.02 °C** of zero, with 95% block-bootstrap intervals spanning zero. Gradients and centered perturbations show a forecast response to the tested inputs. For these cases and this lead, the response to correct timing yielded little detectable accuracy benefit.



*Compare solid with dashed curves for correctly timed and shifted inputs. The temperature-only reference appears in the absolute-error and drop-capture panels.*



*With 16 prior pairs, toggling the warning reduced +24 h q50 by about 0.64 °C. The response was nonmonotonic across counts, each represented by one synthetic history.*

## 6. The short-lead skill advantage fades

Across 127,008 saved Phase I forecasts, the MAE advantage over the same hour yesterday fell from **81.0% at +1 h** to **5.5% at +24 h**. In retrospectively stable conditions, same-hour-yesterday had the lower +24 h MAE, **0.92 °C** versus **1.13 °C**. The sensitivity panels use the separate 144-case Phase II change sample. They show which input slots move forecasts, alongside a separate measure of how accurate those forecasts are.



*The top panels measure error and relative skill. The lower panels show sensitivity magnitude and within-patch concentration.*
