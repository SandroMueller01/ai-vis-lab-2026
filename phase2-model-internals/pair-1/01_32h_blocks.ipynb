#### The question

Exp.10 showed that when we hide one hour of the 4-week history and measure how much the forecast changes, we can notice a consistent 32 hour pattern repeated across the history. This basically means that those 1-hour data points had a stronger impact on the forecast compared to the rest of the history. Since this was computed for random forecasts (random places and dates), the hypothesis that the pattern comes from the data having something more important every 32 hours can be deprioritized, some more input-output experiments could completely rule it out but for time constraints we will focus on the remaining hypothesis: the model reads its history in 32-hour blocks. This finding would point towards a bias in the model structure, since ideally it should set the "relevance" of each context data point only on data-related information.

#### Intuition

The previous setting opens the questions:

+ How does the model exactly decide which hour is more important? or which hour to *pay attention* to?

+ Can we say that hiding an hour and measuring the forecast change is a proper way of defining the importance of that hour?
