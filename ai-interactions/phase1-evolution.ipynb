## 3. Decisions that changed the plan

1. **Smoke test (24 Oct).** Past covariates have no effect, so "use neighbours as inputs" is only an upper bound.
2. **Cold fronts (1 Nov).** A literature search showed that a cold drop is not a front; we added a separate label.
3. **Pair roles (2 Nov).** The 20 Oct plan said Sandro and Carlos are AI-aided in Phase II; the team meeting changed
   that to Phase I. The repo was corrected everywhere.
4. **Password (4 Nov).** A password on the preview was dropped: the data is open (CC BY 4.0) and the repo is public.
5. **Carlos's review of the dashboard.** Three points were fixed (dark only, day picker, readable drop-down).

## 4. What was wrong along the way

- The assistant reported "304 start times plus 32 front runs"; the data has 306 plus 30. Found when the dashboard
  data was built from the file.
- The first "jump to" times used the peak of the front detection, where almost no point was marked.
- A bug dropped the links to Carlos's questions on our charts; a scripted browser test found it.
- A first cross-check of one of Carlos's numbers mixed in the normal evening cooling and had to be redone.
